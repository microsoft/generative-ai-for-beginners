עד כה ראיתם בתכנית הלימודים שיש מושגים מרכזיים כמו הקשות ואפילו תחום שלם הנקרא "הנדסת פרומפטים". כלים רבים שאפשר להשתמש בהם כמו ChatGPT, Office 365, Microsoft Power Platform ועוד, תומכים בשימוש בפרומפטים כדי להשיג מטרה מסוימת.

כדי להוסיף חוויה כזו לאפליקציה, עליכם להבין מושגים כמו פרומפטים, השלמות ולבחור ספרייה לעבודה. זה בדיוק מה שלמדו בפרק זה.

## מבוא

בפרק זה, תלמדו:

- ללמוד על ספריית openai והמושגים המרכזיים שלה.
- לבנות אפליקציית יצירת טקסט באמצעות openai.
- להבין כיצד להשתמש במושגים כמו פרומפט, טמפרטורה וטוקנים כדי לבנות אפליקציית יצירת טקסט.

## מטרות הלמידה

## מהי אפליקציית יצירת טקסט?

בדרך כלל כשאתם בונים אפליקציה יש לה סוג מסוים של ממשק כמו הבא:

- מבוססת פקודות. אפליקציות קונסול הן אפליקציות טיפוסיות שבהן אתם מקלידים פקודה והיא מבצעת משימה. לדוגמה, `git` היא אפליקציה מבוססת פקודות.
- ממשק משתמש (UI). יש אפליקציות עם ממשקי משתמש גרפיים (GUI) שבהם אתם לוחצים על כפתורים, מזינים טקסט, בוחרים אפשרויות ועוד.

### אפליקציות קונסול ו-UI מוגבלות

השוו זאת לאפליקציית קונסול שבה אתם מקלידים פקודה:

- **זה מוגבל**. אינכם יכולים להקליד כל פקודה, רק את אלו שהאפליקציה תומכת בהן.
- **ספציפי לשפה**. יש אפליקציות התומכות בשפות רבות, אך כברירת מחדל האפליקציה בנויה לשפה מסוימת, אפילו אם אפשר להוסיף תמיכה בשפות נוספות.

### יתרונות של אפליקציות יצירת טקסט

אז איך אפליקציית יצירת טקסט שונה?

באפליקציית יצירת טקסט, יש לכם יותר גמישות, אתם לא מוגבלים לקבוצת פקודות או לשפת קלט ספציפית. במקום זאת, אתם יכולים להשתמש בשפה טבעית כדי לתקשר עם האפליקציה. יתרון נוסף הוא שכבר אתם מתקשרים עם מקור נתונים שאומן על קורפוס עצום של מידע, בעוד שאפליקציה מסורתית עלולה להיות מוגבלת למה שקיים בבסיס הנתונים.

### מה אפשר לבנות עם אפליקציית יצירת טקסט?

יש הרבה דברים שאפשר לבנות. לדוגמה:

- **צ'טבוט**. צ'טבוט שעונה על שאלות בנושאים שונים, כמו החברה שלכם והמוצרים שלה, יכול להתאים טוב.
- **עוזר**. מודלים שפתיים גדולים (LLMs) מצוינים במשימות כמו סיכום טקסט, חילוץ תובנות מטקסט, יצירת תוכן כמו קורות חיים ועוד.
- **עוזר קוד**. בהתאם למודל השפה בו אתם משתמשים, ניתן לבנות עוזר קוד שמסייע בכתיבת קוד. לדוגמה, אפשר להשתמש במוצר כמו GitHub Copilot וגם ChatGPT לעזרה בכתיבת קוד.

## איך אפשר להתחיל?

טוב, אתם צריכים למצוא דרך להתחבר למודל שפה גדול (LLM) שלרוב כוללת שתי גישות עיקריות:

- שימוש ב-API. כאן אתם בונים בקשות רשת עם הפרומפט שלכם ומקבלים טקסט שנוצר חזרה.
- שימוש בספרייה. ספריות עוזרות לעטוף את קריאות ה-API ולהפוך אותן לקלות לשימוש.

## ספריות / ערכות פיתוח (SDKs)

יש כמה ספריות מוכרות לעבודה עם LLMs כמו:

בנוסף יש ספריות שפועלות ברמה גבוהה יותר כמו:

## ייצרו טקסט עם ChatCompletions

הדרך לייצר טקסט היא להשתמש במחלקה `ChatCompletionsClient`.
ב-`samples/python/azure_ai_inference/basic.py`, בסעיף התגובה של הקוד, עדכנו את קוד תפקיד המשתמש על ידי שינוי פרמטר התוכן להלן:

```python

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": "Complete the following: Once upon a time there was a",
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

```

הריצו את הקובץ המעודכן כדי לראות את הפלט

1. צור משתנה `prompt` ושנה את קוד הדוגמה כך:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Show me 5 recipes for a dish with the following ingredients: chicken, potatoes, and carrots. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

מעולה, בוא נראה איך נוכל לשפר דברים. כדי לשפר דברים, אנחנו רוצים לוודא שהקוד גמיש, כך שניתן לשפר ולשנות את המרכיבים ומספר המתכונים. 


1. בואו נשנה את הקוד בדרך הבאה:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

no_recipes = input("No of recipes (for example, 5): ")

ingredients = input("List of ingredients (for example, chicken, potatoes, and carrots): ")

# interpolate the number of recipes into the prompt an ingredients
prompt = f"Show me {no_recipes} recipes for a dish with the following ingredients: {ingredients}. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

### 2. Mag-import ng mga tulong na library at gumawa ng mga kredensyal


```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("OPENAI_API_KEY","")
assert API_KEY, "ERROR: OpenAI Key is missing"

client = OpenAI(
    api_key=API_KEY
    )
```

### 3. Paghahanap ng tamang modelo  
Ang mga modelong tulad ng GPT-4o at GPT-4o mini ay kayang makaunawa at makabuo ng natural na wika, at makukuha sa OpenAI platform na may iba't ibang antas ng kapangyarihan at bilis na angkop sa iba't ibang gawain.


```python
# Select a general purpose chat model
model = "gpt-5-mini"
```

## 4. Disenyo ng Prompt  

"Ang mahika ng mga malaking language model ay na sa pamamagitan ng pagsasanay upang mabawasan ang prediction error na ito sa napakalawak na dami ng teksto, natututo ang mga modelo ng mga konseptong kapaki-pakinabang para sa mga prediksyon na ito. Halimbawa, natututo sila ng mga konsepto tulad ng"(1):

#### Paano kontrolin ang isang malaking language model  
"Sa lahat ng input sa isang malaking language model, ang pinaka-maimpluwensiya ay ang text prompt(1).

Maaaring i-prompt ang mga malaking language model upang magproduce ng output sa ilang paraan:

Instruksyon: Sabihin sa modelo kung ano ang gusto mo
Kompletong sagot: Himukin ang modelo na kumpletuhin ang simula ng gusto mo
Demonstrasyon: Ipakita sa modelo kung ano ang gusto mo, sa pamamagitan ng alinman sa:
Ilang halimbawa sa prompt
Maraming daan o libu-libong halimbawa sa isang fine-tuning training dataset"



#### Mayroong tatlong pangunahing gabay sa paggawa ng mga prompt:

**Ipakita at sabihin**. Gawing malinaw kung ano ang gusto mo sa pamamagitan ng mga instruksyon, halimbawa, o kumbinasyon ng dalawa. Kung gusto mong i-ranggo ng modelo ang isang listahan ng mga item ayon sa alpabetikong pagkakasunod o klasipikahin ang isang talata base sa damdamin, ipakita sa modelo na iyon ang gusto mo.

**Magbigay ng kalidad na datos**. Kung sinusubukan mong gumawa ng classifier o gusto mong sundan ng modelo ang isang pattern, tiyaking may sapat na mga halimbawa. Siguraduhing i-proofread ang iyong mga halimbawa — karaniwang matalino ang modelo na makita ang mga pangkaraniwang pagkakamali sa baybay at magbibigay ng sagot, ngunit maaari rin nitong isipin na ito ay sinasadya at maaaring makaapekto sa sagot.

**Suriin ang iyong mga settings.** Kinokontrol ng temperature at top_p settings kung gaano katiyak ang modelo sa paggawa ng sagot. Kung humihingi ka ng sagot kung saan iisa lamang ang tamang sagot, gusto mong ibaba ang mga ito. Kung naghahanap ka ng mas maraming iba't ibang sagot, maaari mong itaas ang mga ito. Ang pinaka-karaniwang pagkakamali ng mga tao sa mga setting na ito ay ang akala nilang ito ay mga kontrol para sa "talino" o "kreatibidad".


Pinagmulan: https://learn.microsoft.com/azure/ai-foundry/openai/overview


### 5. Isumite!


```python
# Create your first prompt
text_prompt = "Should oxford commas always be used?"

response = client.responses.create(
  model=model,
  input = [{"role":"system", "content":"You are a helpful assistant."},
               {"role":"user","content":text_prompt},],
  store=False,)

response.output_text
```

## Buodin ang Teksto  
#### Hamon  
Buodin ang teksto sa pamamagitan ng pagdadagdag ng 'tl;dr:' sa dulo ng isang bahagi ng teksto. Pansinin kung paano nauunawaan ng modelo kung paano isagawa ang maraming gawain nang walang karagdagang mga tagubilin. Maaari kang mag-eksperimento gamit ang mas mapanuring mga prompt kaysa tl;dr upang baguhin ang gawi ng modelo at i-customize ang buod na iyong matatanggap(3).  

Kamakailang mga pag-aaral ang nagpakita ng malalaking pagtaas sa maraming gawain at benchmark sa NLP sa pamamagitan ng pre-training sa isang malaking korpus ng teksto kasunod ang fine-tuning sa isang tiyak na gawain. Bagaman karaniwang hindi naka-angkla sa tiyak na gawain ang arkitektura, nangangailangan pa rin ang metodong ito ng mga dataset ng fine-tuning na tiyak sa gawain ng libo-libo o sampu-sampung libong mga halimbawa. Sa kabilang banda, kadalasan ay kaya ng mga tao na isagawa ang bagong gawain sa wika mula lamang sa ilang mga halimbawa o sa mga simpleng tagubilin - isang bagay na hanggang ngayon ay mahirap gawin ng mga kasalukuyang sistema ng NLP. Dito ipinapakita namin na ang pagpapalawak ng mga modelong pangwika ay lubos na nagpapabuti sa task-agnostic, few-shot na pagganap, na kung minsan ay umaabot pa sa pagiging kumpetisyon sa mga naunang state-of-the-art na pamamaraan ng fine-tuning.  



Tl;dr

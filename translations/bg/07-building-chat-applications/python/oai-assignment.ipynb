### 2. Импортиране на помощни библиотеки и инициализиране на идентификационни данни

#### Как да контролираме голям езиков модел  
"От всички входни данни към голям езиков модел, най-влиятелният по далеч е текстовата подсказка(1).

Големите езикови модели могат да бъдат подканвани да генерират изход по няколко начина:

Инструкция: Кажете на модела какво искате
Допълване: Подтикнете модела да допълни началото на това, което искате
Демонстрация: Покажете на модела какво искате, чрез:
Няколко примера в подсказката
Стотици или хиляди примери в тренировъчен набор за донастройка"



#### Има три основни насоки при създаването на подсказки:

**Покажи и кажи**. Направете ясно какво искате, чрез инструкции, примери или комбинация от двете. Ако искате моделът да подреди списък с елементи по азбучен ред или да класифицира параграф спрямо настроението, покажете му, че това искате.

**Осигурете качествени данни**. Ако се опитвате да създадете класификатор или да насочите модела да следва определен модел, уверете се, че има достатъчно примери. Задължително прегледайте вашите примери — моделът обикновено е достатъчно умен да разпознае основни правописни грешки и да ви даде отговор, но може и да предположи, че това е умишлено и това да повлияе на отговора.

**Проверете настройките си.** Настройките temperature и top_p контролират колко детерминистичен е моделът при генериране на отговор. Ако питате за отговор, където има само един правилен, искате да ги зададете по-ниско. Ако търсите по-разнообразни отговори, може да ги зададете по-високо. Най-честата грешка, която хората допускат с тези настройки, е да мислят, че те контролират "умност" или "креативност".

### 5. Изпратете! 


```python
# Create your first prompt
text_prompt = "Should oxford commas always be used?"

response = client.responses.create(
  model=model,
  input = [{"role":"system", "content":"You are a helpful assistant."},
               {"role":"user","content":text_prompt},],
  store=False,)

response.output_text
```

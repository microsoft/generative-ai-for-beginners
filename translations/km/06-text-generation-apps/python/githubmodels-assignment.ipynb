# បង្កើតកម្មវិធីបង្កើតអត្ថបទ

អ្នកបានឃើញរហូតមកតាមវគ្គសិក្សានេះថាមានគំនិតមូលដ្ឋានដូចជាបញ្ចូលសំណួរ និងមានវិស័យមួយទៀតហៅថា "វិស្វកម្មបញ្ចូលសំណួរ"។ ឧបករណ៍ជាច្រើនដែលអ្នកអាចអន្តរកម្មដូចជា ChatGPT, Office 365, Microsoft Power Platform និងផ្សេងទៀត គាំទ្រអ្នកក្នុងការប្រើបញ្ចូលសំណួរដើម្បីសម្រេចបានអ្វីមួយ។

ដើម្បីឲ្យអ្នកអាចបន្ថែមបទពិសោធន៍ដូចនេះទៅកម្មវិធីមួយ អ្នកត្រូវយល់ពីគំនិតដូចជា បញ្ចូលសំណួរ ការបញ្ចប់ និងជ្រើសយកបណ្ណាល័យមួយសម្រាប់ធ្វើការងារ។ នេះគឺជារឿងដែលអ្នកនឹងរៀននៅក្នុងជំពូកនេះ។

## ការណែនាំ

ក្នុងជំពូកនេះ អ្នកនឹង:

- រៀនអំពីបណ្ណាល័យ openai និងគំនិតស្នូលរបស់វា។
- បង្កើតកម្មវិធីបង្កើតអត្ថបទដោយប្រើ openai។
- យល់អំពីរបៀបប្រើគំនិតដូចជា បញ្ចូលសំណួរ, សីតុណ្ហភាព និងសញ្ញាចូល (tokens) ដើម្បីបង្កើតកម្មវិធីបង្កើតអត្ថបទ។

## គោលបំណងរៀន

នៅចុងវគ្គសិក្សានេះ អ្នកនឹងអាច:

- ពន្យល់ពីអ្វីដែលជាកម្មវិធីបង្កើតអត្ថបទ។
- បង្កើតកម្មវិធីបង្កើតអត្ថបទដោយប្រើ openai។
- កំណត់រចនាសម្ព័ន្ធកម្មវិធីរបស់អ្នកឲ្យប្រើសញ្ញាចូល (tokens) មួយចំនួនច្រើន ឬតិច និងផ្លាស់ប្តូរសីតុណ្ហភាពសម្រាប់លទ្ធផលបម្លែងប្លែកៗ។

## តើកម្មវិធីបង្កើតអត្ថបទជាអ្វី?

ជាទូទៅពេលអ្នកបង្កើតកម្មវិធីវាមានមុខងារមួយប្រភេទដូចខាងក្រោម៖

- ពីការបញ្ជា។ កម្មវិធីកុងសូលជាកម្មវិធីដែលអ្នកវាយបញ្ជា ហើយវាពន្លឿនធ្វើកិច្ចការមួយ។ ឧទាហរណ៍, `git` គឺជាកម្មវិធីពីការបញ្ជា។
- មុខងារអ្នកប្រើ(UI)៖ កម្មវិធីមួយចំនួនមានមុខងារអ្នកប្រើផ្ទាល់ (GUIs) ដែលអ្នកអាចចុចប៊ូតុង បញ្ចូលអត្ថបទ ជ្រើសរើសជម្រើស និងផ្សេងទៀត។

### កម្មវិធីកុងសូលនិង UI មានកំណត់

ប្រៀបធៀបវានឹងកម្មវិធីពីការបញ្ជា ដែលអ្នកវាយបញ្ជា៖

- **វាមានកំណត់**។ អ្នកមិនអាចវាយបញ្ជាណាមួយបានទេ មានតែបញ្ជាដែលកម្មវិធីគាំទ្រតែប៉ុណ្ណោះ។
- **ភាសាផ្សេងៗជាក់លាក់**។ កម្មវិធីមួយចំនួនគាំទ្រភាសាជាច្រើន ប៉ុន្តែជាមុនសិនកម្មវិធីត្រូវបានបង្កើតសម្រាប់ភាសាមួយជាក់លាក់ ទោះបីអ្នកអាចបន្ថែមការគាំទ្រភាសាបានក៏ដោយ។

### អត្ថប្រយោជន៍នៃកម្មវិធីបង្កើតអត្ថបទ

តើកម្មវិធីបង្កើតអត្ថបទខុសពីនេះយ៉ាងដូចម្តេច?

នៅក្នុងកម្មវិធីបង្កើតអត្ថបទ អ្នកមានភាពបត់បែនច្រើនជាងនេះ អ្នកមិនត្រូវបានកំណត់ត្រឹមបញ្ជាមួយកំណត់ ឬភាសាបញ្ចូលជាក់លាក់ទេ។ ផ្ទុយទៅវិញ អ្នកអាចប្រើភាសាធម្មជាតិក្នុងការអន្តរកម្មមកកម្មវិធី។ អត្ថប្រយោជន៍ផ្សេងទៀតគឺ ដោយសារអ្នកបានអន្តរកម្មជាមួយប្រភពទិន្នន័យដែលមានការបណ្តុះបណ្តាលលើសំណុំអត្ថបទធំៗ ខណៈដែលកម្មវិធីប្រពៃណីមួយអាចមានកំណត់លើអ្វីដែលមានក្នុងមូលដ្ឋានទិន្នន័យ។

### តើខ្ញុំអាចបង្កើតអ្វីបានខ្លះជាមួយកម្មវិធីបង្កើតអត្ថបទ?

មានរឿងជាច្រើនដែលអ្នកអាចបង្កើតបាន។ ឧទាហរណ៍៖

- **ប៉ាតបត់**។ ប៉ាតបត់ដែលឆ្លើយសំណួរអំពីប្រធានបទ ឧទាហរណ៍ក្រុមហ៊ុនរបស់អ្នកនិងផលិតផលរបស់វា អាចជាជម្រើសល្អ។
- **ជំនួយការ**។ LLM មានជំនាញល្អក្នុងការសង្ខេបអត្ថបទ, ទាញយកព័ត៌មានពីអត្ថបទ, ផលិតអត្ថបទដូចជា ប្រវត្តិរូប និងផ្សេងទៀត។
- **ជំនួយការសរសេរកូដ**។ អាស្រ័យលើម៉ូដែលភាសាដែលអ្នកប្រើ អ្នកអាចបង្កើតជំនួយការសរសេរកូដដែលជួយអ្នកសរសេរកូដ។ ឧទាហរណ៍ អ្នកអាចប្រើផលិតផលដូចជា GitHub Copilot និង ChatGPT ដើម្បីជួយសរសេរកូដ។

## តើខ្ញុំអាចចាប់ផ្តើមយ៉ាងដូចម្តេច?

អូខេ អ្នកត្រូវស្វែងរកវិធីភ្ជាប់ជាមួយ LLM ដែលភាគច្រើនមានវិធីសាស្រ្តពីរដូចខាងក្រោម៖

- ប្រើ API។ នៅទីនេះ អ្នកកំពុងកសាងការស្នើសុំបណ្ដាញជាមួយបញ្ចូលសំណួរ ហើយទទួលបានអត្ថបទដែលបានបង្កើតវិញ។
- ប្រើបណ្ណាល័យ។ បណ្ណាល័យជួយគ្រប់គ្រងការហៅ API ហើយធ្វើឲ្យវាងាយស្រួលប្រើ។

## បណ្ណាល័យ/SDKs

មានបណ្ណាល័យដែលមានឈ្មោះល្បីច្រើនសម្រាប់ធ្វើការជាមួយ LLM ដូចជា:

- **openai**, បណ្ណាល័យនេះធ្វើឲ្យការតភ្ជាប់ទៅម៉ូដែលរបស់អ្នក និងផ្ញើបញ្ចូលសំណួរជាងមានភាពងាយស្រួល។

បន្ទាប់មកមានបណ្ណាល័យដែលដំណើរការលើកម្រិតខ្ពស់ជាងនេះដូចជា:

- **Langchain**. Langchain មានឈ្មោះល្បី និងគាំទ្រភាសា Python។
- **Semantic Kernel**. Semantic Kernel គឺជាបណ្ណាល័យមួយដោយ Microsoft ដែលគាំទ្រភាសា C#, Python, និង Java។

## បង្កើតអត្ថបទជាមួយ ChatCompletions

នៅក្នុង `samples/python/azure_ai_inference/basic.py` នៅផ្នែកនៃការឆ្លើយតប នៃកូដ បន្ថែមអ្នកប្រើដោយផ្លាស់ប្ដូរពាក្យពណ៌នា content ដូចខាងក្រោម៖

```python

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": "Complete the following: Once upon a time there was a",
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

```

១។ ប្រើឯកសារដែលមានស្រាប់ជារូបមន្តចាប់ផ្តើម
១។ បង្កើតអថេរ `prompt` ហើយផ្លាស់ប្តូរកូដគំរូដូចខាងក្រោម៖


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Show me 5 recipes for a dish with the following ingredients: chicken, potatoes, and carrots. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

no_recipes = input("No of recipes (for example, 5): ")

ingredients = input("List of ingredients (for example, chicken, potatoes, and carrots): ")

# interpolate the number of recipes into the prompt an ingredients
prompt = f"Show me {no_recipes} recipes for a dish with the following ingredients: {ingredients}. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

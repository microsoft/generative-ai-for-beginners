# ការសាងសង់កម្មវិធីបង្កើតរូបភាព (OpenAI)

មានអ្វីខុសពីការបង្កើតអត្ថបទនៃ LLMs ។ អ្នកអាចបង្កើតរូបភាពពីការពិពណ៌នាអត្ថបទផងដែរ។ រូបភាពជារបៀបប្រើមានប្រយោជន៍នៅក្នុងវិស័យម៉េដិច, សំណង់, ទេសចរណ៍, ការអភិវឌ្ឍហ្គេម, ផ្សព្វផ្សាយ, និងផ្សេងទៀត។ ក្នុងមេរៀននេះ យើងបំពេញជាមួយម៉ូឌែល **GPT Image** សព្វថ្ងៃនៅលើវេទិកា OpenAI ។

## គោលបំណងសិក្សា

នៅចប់មេរៀននេះ អ្នកនឹងអាច:

- បកស្រាយថាការបង្កើតរូបភាពមានអ្វី និងវាមានប្រយោជន៍នៅទីណា។
- យល់ពីគ្រួសារ ម៉ូឌែល `gpt-image` និងរបៀបវាផ្សេងពីម៉ូឌែល DALL·E ចាស់។
- បង្កើតកម្មវិធីបង្កើតរូបភាព និងកែសម្រួលរូបភាព។

## ការបង្កើតរូបភាពគឺជាម្ដេច?

ម៉ូឌែលបង្កើតរូបភាពបង្កើតរូបភាពពីការបញ្ជាក់អត្ថបទ។ ម៉ូឌែលទំនើបដូចជា `gpt-image` យល់ពីទំនាក់ទំនងរវាងអត្ថបទនិងរូបភាពក្នុងដំណាក់កាលបណ្តុះបណ្តាល បន្ទាប់មកបំលែងសំឡេងច្រឡំឱ្យទៅជារូបភាពដែលត្រូវគ្នានឹងបញ្ជាក់របស់អ្នកជាទំនៀមទន្ទឹម។

ពីរគ្រួសារល្បីល្បាញនៃម៉ូឌែលរូបភាពមាន៖

- **`gpt-image` (OpenAI)** — ជំនាន់បច្ចុប្បន្នដែលប្រើក្នុងមេរៀននេះ។ វាសម្របសម្រួលការបង្កើតរូបភាពពីអត្ថបទ និងកែសម្រួលរូបភាព (inpainting ជាមួយម៉ាស)។
- **Midjourney** — ម៉ូឌែលភាគីទីបីដែលពេញនិយមជាមួយសេវាកម្មខ្លួនឯង និងបច្ចេកវិទ្យាការងារចល័តលើ Discord។

> ម៉ូឌែលរូបភាព OpenAI ចាស់ — **DALL·E 2** និង **DALL·E 3** — គឺជាម៉ូឌែលចាស់។ DALL·E 3 មិនមានសម្រាប់ការដាក់ឱ្យដំណើរការថ្មីទៀតទេ ហើយមុខងារ​ដូចជា `create_variation` មានតែ​នៅក្នុង DALL·E 2 ប៉ុណ្ណោះ។ ប្រើម៉ូឌែល `gpt-image` សម្រាប់កម្មវិធីថ្មី។

> **យ៉ាងសំខាន់ៈ** ម៉ូឌែល `gpt-image` បញ្ជូនរូបភាពដែលបានបង្កើតជាសង្វាក់ **base64** (`b64_json`) មិនមែនជាចំណុច URL ទេ។ កូដរបស់អ្នកបកប្រែខ្សែ base64 ទៅជាអក្សរបាញ់ bytes ហើយរក្សាទុកវា — មិនមាន URL រូបភាពអាចទាញយកបានទេ។


## ការបង្កើតកម្មវិធីបង្កើតរូបភាពដំបូងរបស់អ្នក

ហេតុអ្វីបានជា ត្រូវការអ្វីខ្លះដើម្បីបង្កើតកម្មវិធីបង្កើតរូបភាព? អ្នកត្រូវការជំនួយបណ្ណាល័យដូចតទៅ៖

- **python-dotenv**, អ្នកត្រូវបានណែនាំយ៉ាងខ្លាំងឲ្យប្រើ​បណ្ណាល័យនេះ ដើម្បីរក្សារការសម្ងាត់របស់អ្នកក្នុងឯកសារ *.env* ដែលឆ្ងាយពីកូដ។
- **openai**, បណ្ណាល័យនេះគឺជាអ្វីដែលអ្នកនឹងប្រើដើម្បីធ្វើអន្តរកម្មជាមួយ OpenAI API។
- **pillow**, ដើម្បីធ្វើការជាមួយរូបភាពនៅក្នុង Python។


1. បង្កើតឯកសារ *.env* ជាមួយមាតិកាតទៅខាងក្រោម៖

    ```text
    OPENAI_API_KEY='<add your OpenAI key here>'
    ```



1. ប្រមូលបណ្ណាល័យខាងលើ​ទាំងអស់​ក្នុង​ឯកសារ​មួយ​ឈ្មោះ *requirements.txt* ដូច្នេះ៖

    ```text
    python-dotenv
    openai
    pillow
    ```


1. បន្ទាប់មក បង្កើត​បរិយាកាស​ស្មើរដូច​មុខ និងដំឡើង​បណ្ណាល័យ​ទាំងនោះ៖


`````python
# create virtual env
! python3 -m venv venv
# activate environment
! source venv/bin/activate
# install libraries
# pip install -r requirements.txt, if using a requirements.txt file 
! pip install python-dotenv openai pillow
`````

> [!NOTE]
> សម្រាប់ Windows សូមប្រើពាក្យបញ្ជាខាងក្រោមដើម្បីបង្កើតនិងបើកការបរិស្ថានមេរោគរបស់អ្នក៖

    ```bash
    python3 -m venv venv
    venv\Scripts\activate.bat
    ````

1. បន្ថែមកូដខាងក្រោមនៅក្នុងឯកសារដែលហៅថា *app.py*:

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    import openai

    # import dotenv
    dotenv.load_dotenv()

    # Create OpenAI object (reads OPENAI_API_KEY from your .env)
    client = openai.OpenAI()


    try:
        # Create an image by using the image generation API
        generation_response = client.images.generate(
            model="gpt-image-2",
            prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # Enter your prompt text here
            size='1024x1024',
            n=1
        )
        # Set the directory for the stored image
        image_dir = os.path.join(os.curdir, 'images')

        # If the directory doesn't exist, create it
        if not os.path.isdir(image_dir):
            os.mkdir(image_dir)

        # Initialize the image path (note the filetype should be png)
        image_path = os.path.join(image_dir, 'generated-image.png')

        # gpt-image models return the image as base64 (b64_json), not a URL
        image_b64 = generation_response.data[0].b64_json
        generated_image = base64.b64decode(image_b64)
        with open(image_path, "wb") as image_file:
            image_file.write(generated_image)

        # Display the image in the default image viewer
        image = Image.open(image_path)
        image.show()

    # catch exceptions
    except openai.BadRequestError as err:
        print(err)

    ```

យើងចូលចិត្តបកស្រាយកូដនេះ៖

- ជាដំបូង យើងនាំចូលបណ្ណាល័យដែលយើងត្រូវការជាមួយនឹងបណ្ណាល័យ OpenAI, បណ្ណាល័យ dotenv, ម៉ូឌុល base64, និងបណ្ណាល័យ Pillow។

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    import openai
    ```

- បន្ទាប់មក យើងបង្កើត client ដែលអាន API key ពី ``.env`` របស់អ្នក។

    ```python
    # Create OpenAI object
    client = openai.OpenAI()
    ```

- បន្ទាប់ មក យើងបង្កើតរូបភាព៖

    ```python
    generation_response = client.images.generate(
        model="gpt-image-2",
        prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # Enter your prompt text here
        size='1024x1024',
        n=1
    )
    ```

    `gpt-image` ម៉ូឌែលបញ្ចូនរូបភាពជាខ្សែអក្សរ **base64** នៅក្នុង `data[0].b64_json`។ យើងបញ្ចេញវាទៅជាបៃនិងសរសេរវាទៅឯកសារ — មិនមាន URL សម្រាប់ទាញយកទេ។

    ```python
    image_b64 = generation_response.data[0].b64_json
    generated_image = base64.b64decode(image_b64)
    with open(image_path, "wb") as image_file:
        image_file.write(generated_image)
    ```

- ចុងក្រោយ យើងបើករូបភាពនិងប្រើកម្មវិធីមើលរូបភាពស្តង់ដារដើម្បីបង្ហាញវា៖

    ```python
    image = Image.open(image_path)
    image.show()
    ```

### ព័ត៌មានលម្អិតបន្ថែមអំពីការបង្កើតរូបភាព

មកមើលប៉ារ៉ាម៉ែត្រនៃ `images.generate`៖

- **model**, គឺជាម៉ូឌែលរូបភាពដែលត្រូវប្រើ (ឧទាហរណ៍ `gpt-image-2`)។
- **prompt**, គឺជាអត្ថបទប្រើសម្រាប់បង្កើតរូបភាព។ នៅទីនេះគឺ "Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils"។
- **size**, គឺជា​ទំហំរូបភាពដែលបានបង្កើត (ឧទាហរណ៍ `1024x1024`, `1536x1024`, `1024x1536`, ឬ `"auto"`)។
- **n**, គឺជាចំនួនរូបភាពដែលបានបង្កើត។ នៅទីនេះយើងបង្កើតមួយគត់។

> ម៉ូឌែលរូបភាពមិនទទួលយកប៉ារ៉ាម៉ែត្រ `temperature` ទេ — នេះគឺជាការត្រួតពិនិត្យការបង្កើតអត្ថបទ។ ដើម្បីមានភាពចម្រុះ សូមហៅ API ម្តងទៀត; ដើម្បីកាត់បន្ថយភាពចម្រុះ សូមធ្វើអោយ prompt របស់អ្នកជាក់លាក់ជាង។

## សមត្ថភាព​បន្ថែមនៃការបង្កើតរូបភាព

អ្នកបានឃើញរបៀបបង្កើតរូបភាពមួយជាមួយបន្ទាត់កូដ Python តិចតួចមួយហើយ។ ម៉ូដែល `gpt-image` ក៏អាច **កែប្រែ** រូបភាពដែលមានស្រាប់បានផងដែរ។ ដោយផ្តល់រូបភាពមួយ ក្បាលបុព្វបំណង **ម៉ាស** (mask) ជាជម្រើសដែលសម្គាល់តំបន់ដែលត្រូវបម្លែង និងបទបញ្ជា ត្រូវអាចប្ដូរផ្នែកមួយនៃរូបភាព — ឧទាហរណ៍ បន្ថែមមួកឲ្យកូនខ្លាញ់របស់យើង។

```python
response = client.images.edit(
  model="gpt-image-2",
  image=open("base_image.png", "rb"),
  mask=open("mask.png", "rb"),
  prompt="An image of a rabbit with a hat on its head.",
  n=1,
  size="1024x1024"
)
# edits are also returned as base64
edited_image = base64.b64decode(response.data[0].b64_json)
```

រូបភាពគោលមានតែខ្លាច់; រូបភាពចុងក្រោយមានមួកលើខ្លាច់។

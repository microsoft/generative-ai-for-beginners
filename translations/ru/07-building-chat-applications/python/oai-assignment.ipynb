### 2. Импорт вспомогательных библиотек и создание учетных данных


```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("OPENAI_API_KEY","")
assert API_KEY, "ERROR: OpenAI Key is missing"

client = OpenAI(
    api_key=API_KEY
    )
```

### 3. Поиск подходящей модели  
Модели, такие как GPT-4o и GPT-4o mini, могут понимать и генерировать естественный язык, и доступны на платформе OpenAI с разными уровнями мощности и скорости, подходящими для различных задач.


```python
# Select a general purpose chat model
model = "gpt-5-mini"
```

## 4. Проектирование подсказок  

«Магия больших языковых моделей заключается в том, что, обучаясь минимизировать ошибку предсказания на огромных объемах текста, модели в итоге начинают усваивать концепции, полезные для этих предсказаний. Например, они учат такие концепции, как»(1):

* как правильно писать
* как работает грамматика

* как отвечать на вопросы
* как вести беседу
* как писать на многих языках
* как программировать
* и т. д.

#### Как управлять большой языковой моделью  
«Из всех входных данных большой языковой модели самой влиятельной, безусловно, является текстовая подсказка»(1).

Большие языковые модели можно подтолкнуть к генерации вывода несколькими способами:

Инструкция: Скажите модели, чего вы хотите
Завершение: Побудите модель завершить начало того, что вы хотите
Демонстрация: Покажите модели, чего вы хотите, с помощью либо:
Нескольких примеров в подсказке
Сотен или тысяч примеров в тренировочном наборе для тонкой настройки»



#### Есть три основных правила создания подсказок:

**Показывайте и объясняйте**. Сделайте понятно, чего вы хотите, через инструкции, примеры или их комбинацию. Если вы хотите, чтобы модель упорядочила список элементов в алфавитном порядке или классифицировала абзац по эмоциональному окрасу, покажите ей, что именно вы хотите.

**Обеспечьте качественные данные**. Если вы пытаетесь создать классификатор или заставить модель следовать определенному шаблону, убедитесь, что примеров достаточно. Обязательно проверьте ваши примеры — модель обычно достаточно умна, чтобы видеть базовые орфографические ошибки и дать ответ, но она также может предположить, что это сделано намеренно, и это может повлиять на ответ.

**Проверьте ваши настройки.** Параметры temperature и top_p контролируют, насколько детерминированной будет модель при генерации ответа. Если вы ожидаете ответ, где есть только один правильный вариант, стоит установить их ниже. Если вы хотите получить более разнообразные ответы, установите их выше. Главная ошибка, которую люди делают с этими настройками — думать, что они контролируют «умность» или «креативность».


Источник: https://learn.microsoft.com/azure/ai-foundry/openai/overview


### 5. Отправить!


```python
# Create your first prompt
text_prompt = "Should oxford commas always be used?"

response = client.responses.create(
  model=model,
  input = [{"role":"system", "content":"You are a helpful assistant."},
               {"role":"user","content":text_prompt},],
  store=False,)

response.output_text
```

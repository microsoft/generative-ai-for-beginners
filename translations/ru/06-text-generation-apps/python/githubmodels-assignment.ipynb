## Что такое приложение для генерации текста?

Обычно при создании приложения оно имеет какой-то интерфейс, например такой:

- Командный. Консольные приложения – типичные примеры, где вы вводите команду, и она выполняет задачу. Например, `git` — это приложение с командным интерфейсом.
- Пользовательский интерфейс (UI). Некоторые приложения имеют графический интерфейс (GUI), где вы нажимаете кнопки, вводите текст, выбираете опции и так далее.

### Консольные и интерфейсные приложения ограничены

Сравните с командным приложением, где вы вводите команду:

- **Ограничено**. Вы не можете просто так ввести любую команду, только те, которые поддерживаются приложением.
- **Языковая специфика**. Некоторые приложения поддерживают несколько языков, но по умолчанию приложение создаётся под конкретный язык, даже если можно добавить поддержку других.

### Преимущества приложений генерации текста

Так чем же приложение генерации текста отличается?

В таких приложениях больше гибкости: вы не ограничены набором команд или конкретным языком ввода. Вместо этого вы используете естественный язык для взаимодействия с приложением. Ещё одно преимущество в том, что вы работаете с источником данных, который обучен на огромном корпусе информации, тогда как традиционное приложение ограничено тем, что есть в базе данных.

### Что можно создать с помощью приложения генерации текста?

## Упражнение — создать генератор рецептов

Теперь, когда мы проиграли сценарий, давайте напишем код, соответствующий продемонстрированному сценарию. Для этого выполните следующие шаги:

1. Используйте существующий файл в качестве отправной точки
1. Создайте переменную `prompt` и измените пример кода следующим образом:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Show me 5 recipes for a dish with the following ingredients: chicken, potatoes, and carrots. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

Отлично, давайте посмотрим, как мы можем улучшить ситуацию. Чтобы улучшить ситуацию, мы хотим, чтобы код был гибким, чтобы ингредиенты и количество рецептов можно было улучшать и изменять.


1. Давайте изменим код следующим образом:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

no_recipes = input("No of recipes (for example, 5): ")

ingredients = input("List of ingredients (for example, chicken, potatoes, and carrots): ")

# interpolate the number of recipes into the prompt an ingredients
prompt = f"Show me {no_recipes} recipes for a dish with the following ingredients: {ingredients}. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

### 2. Hilfsbibliotheken importieren und Anmeldedaten instanziieren


```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("OPENAI_API_KEY","")
assert API_KEY, "ERROR: OpenAI Key is missing"

client = OpenAI(
    api_key=API_KEY
    )
```

### 3. Das richtige Modell finden  
Modelle wie GPT-4o und GPT-4o mini können natürliche Sprache verstehen und erzeugen und sind auf der OpenAI-Plattform mit unterschiedlichen Leistungsstufen und Geschwindigkeiten verfügbar, die für verschiedene Aufgaben geeignet sind.


```python
# Select a general purpose chat model
model = "gpt-5-mini"
```

## 4. Prompt-Design  

"Die Magie großer Sprachmodelle besteht darin, dass sie, indem sie darauf trainiert werden, diesen Vorhersagefehler über enorme Textmengen zu minimieren, letztendlich Konzepte lernen, die für diese Vorhersagen nützlich sind. Zum Beispiel lernen sie Konzepte wie"(1):

* wie man buchstabiert

* wie man ein Gespräch führt
* wie man in vielen Sprachen schreibt
* wie man programmiert
* usw.

#### Wie man ein großes Sprachmodell steuert  
"Von allen Eingaben an ein großes Sprachmodell ist bei weitem der einflussreichste der Texteingabe-Prompt(1).

Große Sprachmodelle können auf verschiedene Weise dazu gebracht werden, eine Ausgabe zu erzeugen:

Anweisung: Sage dem Modell, was du möchtest
Vervollständigung: Veranlasse das Modell, den Anfang dessen zu vervollständigen, was du möchtest
Demonstration: Zeige dem Modell, was du möchtest, entweder durch:
Einige Beispiele im Prompt
Hunderte oder Tausende von Beispielen in einem Feinabstimmungs-Trainingsdatensatz"



#### Es gibt drei grundlegende Richtlinien zur Erstellung von Prompts:

**Zeigen und sagen**. Mache klar, was du möchtest, entweder durch Anweisungen, Beispiele oder eine Kombination aus beidem. Wenn du möchtest, dass das Modell eine Liste von Elementen alphabetisch sortiert oder einen Absatz nach Stimmung klassifiziert, zeige ihm, dass das dein Wunsch ist.

**Stelle qualitativ hochwertige Daten zur Verfügung**. Wenn du versuchst, einen Klassifikator zu erstellen oder das Modell einem Muster folgen soll, stelle sicher, dass genügend Beispiele vorhanden sind. Überprüfe deine Beispiele sorgfältig – das Modell ist in der Regel schlau genug, grundlegende Rechtschreibfehler zu erkennen und dir eine Antwort zu geben, aber es könnte auch annehmen, dass dies absichtlich ist, was die Antwort beeinflussen kann.

**Überprüfe deine Einstellungen.** Die Parameter temperature und top_p steuern, wie deterministisch das Modell bei der Generierung einer Antwort ist. Wenn du eine Antwort möchtest, bei der es nur eine richtige Lösung gibt, solltest du diese niedriger einstellen. Wenn du vielfältigere Antworten suchst, kannst du sie höher einstellen. Der häufigste Fehler bei diesen Einstellungen ist die Annahme, sie seien Steuerungen für „Klugheit“ oder „Kreativität“.


Quelle: https://learn.microsoft.com/azure/ai-foundry/openai/overview


### 5. Einreichen!


```python
# Create your first prompt
text_prompt = "Should oxford commas always be used?"

response = client.responses.create(
  model=model,
  input = [{"role":"system", "content":"You are a helpful assistant."},
               {"role":"user","content":text_prompt},],
  store=False,)

response.output_text
```

# สร้างแอปสร้างข้อความ

คุณได้เห็นจนถึงตอนนี้ในหลักสูตรนี้ว่ามีแนวคิดหลักเช่น prompt และแม้แต่วินัยทั้งหมดที่เรียกว่า "วิศวกรรม prompt" เครื่องมือต่าง ๆ ที่คุณสามารถโต้ตอบด้วยได้ เช่น ChatGPT, Office 365, Microsoft Power Platform และอื่น ๆ รองรับการใช้ prompt เพื่อทำสิ่งต่าง ๆ ให้สำเร็จ

เพื่อให้คุณสามารถเพิ่มประสบการณ์ดังกล่าวลงในแอป คุณจำเป็นต้องเข้าใจแนวคิดอย่าง prompt, completion และเลือกไลบรารีที่จะทำงานด้วย ซึ่งนี่คือสิ่งที่คุณจะได้เรียนรู้ในบทนี้

## บทนำ

ในบทนี้คุณจะ:

- เรียนรู้เกี่ยวกับไลบรารี openai และแนวคิดหลักของมัน
- สร้างแอปสร้างข้อความโดยใช้ openai
- เข้าใจวิธีใช้แนวคิดเช่น prompt, temperature, และ tokens เพื่อสร้างแอปสร้างข้อความ

## เป้าหมายการเรียนรู้

เมื่อจบบทเรียนนี้ คุณจะสามารถ:

- อธิบายว่าแอปสร้างข้อความคืออะไร
- สร้างแอปสร้างข้อความโดยใช้ openai
- กำหนดค่าแอปของคุณให้ใช้ tokens มากหรือน้อย และเปลี่ยนอุณหภูมิ เพื่อให้ได้ผลลัพธ์ที่หลากหลาย

## แอปสร้างข้อความคืออะไร?

โดยปกติเมื่อคุณสร้างแอป มันจะมีอินเทอร์เฟซบางอย่างเหมือนต่อไปนี้:

- แบบใช้คำสั่ง แอปคอนโซลเป็นแอปประเภทที่คุณพิมพ์คำสั่งและจะดำเนินการตามงาน เช่น `git` เป็นแอปแบบใช้คำสั่ง
- ส่วนติดต่อผู้ใช้ (UI) แอปบางตัวมีส่วนติดต่อผู้ใช้แบบกราฟิก (GUI) ที่คุณคลิกปุ่ม ป้อนข้อความ เลือกตัวเลือก และอื่น ๆ

### แอปคอนโซลและ UI มีข้อจำกัด

เปรียบเทียบกับแอปแบบใช้คำสั่งที่คุณพิมพ์คำสั่ง:

- **มีข้อจำกัด** คุณไม่สามารถพิมพ์คำสั่งอะไรก็ได้ มีเพียงคำสั่งที่แอปรองรับเท่านั้น
- **จำกัดเฉพาะภาษา** แอปบางตัวรองรับหลายภาษา แต่โดยปกติแล้วแอปจะถูกสร้างมาเพื่อภาษาหนึ่งโดยเฉพาะ แม้ว่าคุณจะเพิ่มการรองรับภาษาอื่นได้

### ประโยชน์ของแอปสร้างข้อความ

แล้วแอปสร้างข้อความแตกต่างอย่างไร?

ในแอปสร้างข้อความ คุณมีความยืดหยุ่นมากขึ้น ไม่ถูกจำกัดด้วยชุดคำสั่งหรือภาษาอินพุตเฉพาะ แต่คุณสามารถใช้ภาษาธรรมชาติเพื่อโต้ตอบกับแอป อีกประโยชน์หนึ่งคือเพราะคุณกำลังโต้ตอบกับแหล่งข้อมูลที่ได้รับการฝึกฝนจากข้อมูลจำนวนมาก ขณะที่แอปแบบดั้งเดิมอาจถูกจำกัดด้วยข้อมูลที่เก็บในฐานข้อมูล

### ฉันสามารถสร้างอะไรได้บ้างด้วยแอปสร้างข้อความ?

คุณสามารถสร้างได้หลายอย่าง เช่น:

- **แชทบอท** แชทบอทตอบคำถามเกี่ยวกับหัวข้อต่าง ๆ เช่นบริษัทของคุณและผลิตภัณฑ์ เป็นตัวอย่างที่ดี
- **ผู้ช่วย** LLMs เหมาะสำหรับงานอย่างการสรุปข้อความ, หาข้อมูลเชิงลึกจากข้อความ, ผลิตข้อความเช่นประวัติย่อ และอื่น ๆ
- **ผู้ช่วยเขียนโค้ด** ขึ้นอยู่กับโมเดลภาษาที่คุณใช้ คุณสามารถสร้างผู้ช่วยเขียนโค้ด ที่ช่วยคุณเขียนโค้ดได้ เช่น ใช้ผลิตภัณฑ์อย่าง GitHub Copilot รวมถึง ChatGPT เพื่อช่วยเขียนโค้ด

## ฉันจะเริ่มต้นอย่างไร?

คุณต้องหาวิธีเชื่อมต่อกับ LLM ซึ่งมักจะมีสองวิธีนี้:

- ใช้ API ที่นี่คุณสร้างคำขอเว็บด้วยพรอมต์ของคุณและรับข้อความที่ถูกสร้างกลับมา
- ใช้ไลบรารี ไลบรารีช่วยห่อหุ้มการเรียก API และทำให้ง่ายต่อการใช้งาน

## ไลบรารี/SDKs

มีไลบรารีที่รู้จักดีบางตัวสำหรับทำงานกับ LLM เช่น:

- **openai** ไลบรารีนี้ช่วยให้เชื่อมต่อกับโมเดลของคุณและส่งพรอมต์ได้ง่าย

และก็มีไลบรารีที่ทำงานในระดับสูงกว่า เช่น:

- **Langchain** Langchain เป็นที่รู้จักและรองรับ Python
- **Semantic Kernel** Semantic Kernel เป็นไลบรารีของ Microsoft รองรับภาษาต่าง ๆ เช่น C#, Python และ Java

## สร้างข้อความด้วย ChatCompletions

วิธีสร้างข้อความคือใช้คลาส `ChatCompletionsClient`
ใน `samples/python/azure_ai_inference/basic.py` ในส่วนโค้ด response ให้แก้ไขบทบาทผู้ใช้โดยเปลี่ยนพารามิเตอร์ content เป็นดังนี้:

```python

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": "Complete the following: Once upon a time there was a",
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

```

รันไฟล์ที่อัปเดตเพื่อดูผลลัพธ์


## ประเภทของพรอมต์ที่แตกต่างกัน สำหรับสิ่งที่แตกต่างกัน

ตอนนี้คุณได้เห็นวิธีสร้างข้อความโดยใช้พรอมต์แล้ว คุณยังมีโปรแกรมที่พร้อมใช้งานซึ่งคุณสามารถแก้ไขและปรับเปลี่ยนเพื่อสร้างข้อความประเภทต่าง ๆ ได้

พรอมต์สามารถใช้สำหรับงานต่าง ๆ ได้มากมาย ตัวอย่างเช่น:

- **สร้างข้อความประเภทหนึ่ง** ตัวอย่างเช่น คุณสามารถสร้างบทกวี คำถามสำหรับแบบทดสอบ ฯลฯ
- **ค้นหาข้อมูล** คุณสามารถใช้พรอมต์เพื่อค้นหาข้อมูลเช่นตัวอย่างนี้ 'CORS หมายความว่าอย่างไรในการพัฒนาเว็บ?'
- **สร้างโค้ด** คุณสามารถใช้พรอมต์เพื่อสร้างโค้ด เช่น การพัฒนากฎนิพจน์ปกติ (regular expression) ใช้ตรวจสอบอีเมล หรือจะสร้างโปรแกรมทั้งโปรแกรมอย่างเว็บแอปก็ได้

## แบบฝึกหัด - สร้างเครื่องมือสร้างสูตรอาหาร

ตอนนี้ที่เราได้เล่นสถานการณ์ไปแล้ว ให้เขียนโค้ดให้ตรงกับสถานการณ์ที่แสดงไว้ ในการทำเช่นนั้น ให้ทำตามขั้นตอนเหล่านี้:

1. ใช้ไฟล์ที่มีอยู่เป็นจุดเริ่มต้น
1. สร้างตัวแปร `prompt` และเปลี่ยนโค้ดตัวอย่างดังนี้:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Show me 5 recipes for a dish with the following ingredients: chicken, potatoes, and carrots. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

ดีมาก มาดูกันว่าเราจะปรับปรุงอะไรได้บ้าง เพื่อปรับปรุงสิ่งต่างๆ เราต้องการให้โค้ดมีความยืดหยุ่น เพื่อวัตถุดิบและจำนวนสูตรอาหารสามารถปรับปรุงและเปลี่ยนแปลงได้


1. มาเปลี่ยนโค้ดในแบบต่อไปนี้กันเถอะ:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

no_recipes = input("No of recipes (for example, 5): ")

ingredients = input("List of ingredients (for example, chicken, potatoes, and carrots): ")

# interpolate the number of recipes into the prompt an ingredients
prompt = f"Show me {no_recipes} recipes for a dish with the following ingredients: {ingredients}. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

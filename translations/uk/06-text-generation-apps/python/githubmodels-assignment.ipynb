## Що таке додаток для генерації тексту?

Зазвичай, коли ви створюєте додаток, він має якийсь інтерфейс, наприклад, такий:

- Командний інтерфейс. Консольні додатки - це типовий приклад, коли ви вводите команду, і вона виконує завдання. Наприклад, `git` — це додаток із командним інтерфейсом.
- Користувацький інтерфейс (UI). Деякі додатки мають графічний інтерфейс користувача (GUI), де ви натискаєте кнопки, вводите текст, вибираєте опції й інше.

### Консольні та UI-додатки мають обмеження

Порівняйте це з додатком на основі команд, де ви вводите команду:

- **Він обмежений**. Ви не можете просто так ввести будь-яку команду, лише ті, що підтримує додаток.
- **Мовоспецифічний**. Деякі додатки підтримують багато мов, але за замовчуванням додаток розроблений для конкретної мови, навіть якщо ви можете додати підтримку інших.

### Переваги додатків для генерації тексту

Отже, чим же відрізняється додаток для генерації тексту?

У додатку для генерації тексту у вас є більше гнучкості, ви не обмежені набором команд або конкретною мовою вводу. Натомість ви можете використовувати природну мову для взаємодії з додатком. Інша перевага полягає в тому, що ви вже взаємодієте з джерелом даних, яке було навчено на величезному корпусі інформації, тоді як традиційний додаток може мати обмеження в тому, що є в базі даних.

### Що я можу створити з додатком для генерації тексту?

Ви можете створити багато речей, наприклад:

- **Чатбот**. Чатбот, який відповідає на запитання про теми, такі як ваша компанія та її продукти, може бути хорошим варіантом.
- **Помічник**. Моделі великих мов (LLM) чудово справляються із завданнями, як-от підсумовування тексту, отримання інсайтів з тексту, створення текстів, наприклад резюме, і не тільки.
- **Помічник з кодування**. Залежно від моделі мови, яку ви використовуєте, ви можете створити помічника, який допомагає писати код. Наприклад, ви можете скористатися такими продуктами, як GitHub Copilot і ChatGPT, для підтримки у написанні коду.

## Вправа - створити генератор рецептів

Тепер, коли ми програли сценарій, давайте напишемо код, щоб відповідати показаному сценарію. Для цього виконайте наступні кроки:

1. Використайте існуючий файл як початкову точку
1. Створіть змінну `prompt` і змініть приклад коду як нижче:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Show me 5 recipes for a dish with the following ingredients: chicken, potatoes, and carrots. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

1. Давайте змінимо код таким чином:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

no_recipes = input("No of recipes (for example, 5): ")

ingredients = input("List of ingredients (for example, chicken, potatoes, and carrots): ")

# interpolate the number of recipes into the prompt an ingredients
prompt = f"Show me {no_recipes} recipes for a dish with the following ingredients: {ingredients}. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

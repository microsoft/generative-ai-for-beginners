# ပုံများ ဖန်တီးရေး အက်ပ်လီကေးရှင်းများ ဆောက်လုပ်ခြင်း (OpenAI)

LLM များမှာ စာသားဖန်တီးခြင်းထက် ပိုမိုစွမ်းဆောင်နိုင်သည်။ သင်သည် စာသားဖော်ပြချက်များမှ ပုံများကိုလည်း ဖန်တီးနိုင်သည်။ ပုံများသည် MedTech, ဘုပျိုးပညာ၊ سياحت၊ ဂိမ်းထွန်းခြင်း၊ စျေးကွက်ရှာဖွေရေးနှင့် အမျိုးမျိုးသောကဏ္ဍများတွင် အသုံးဝင်သည်။ ဒီသင်ခန်းစာတွင် ကျွန်ုပ်တို့သည် OpenAI ပလက်ဖောင်းပေါ်ရှိ နောက်ဆုံးထွက် **GPT Image** မော်ဒယ်များနှင့် လုပ်ဆောင်ကြသည်။

## သင်ယူရမည့် ရည်ရွယ်ချက်များ

ဒီသင်ခန်းစာကုန်ဆုံးတဲ့အချိန်မှာ သင့်မှာ ပြုလုပ်နိုင်မှာတွေက:

- ပုံဖန်တီးခြင်း ဆိုသည်မှာ အဘယ်နည်းနှင့် ဘယ်နေရာများတွင် အသုံးဝင်ကြောင်း ရှင်းပြနိုင်မည်။
- `gpt-image` မော်ဒယ် မိသားစုနှင့် ဒါဟာ အရင်က DALL·E မော်ဒယ်များနှင့် မည်သို့ ကွဲပြားသည်ကို နားလည်နိုင်မည်။
- ပုံဖန်တီးရေးအက်ပ်ဖြစ်သည့် အက်ပ်တစ်ခု တည်ဆောက်ပြီး ပုံများကို တည်းဖြတ်နိုင်မည်။

## ပုံဖန်တီးခြင်း ဆိုသည်မှာ အဘယ်နည်း?

ပုံဖန်တီးမော်ဒယ်များသည် စာသားဖော်ပြချက်မှ ပုံများကို ဖန်တီးပေးသည်။ ခေတ်မွန်းမိုး မော်ဒယ်များဖြစ်သည့် `gpt-image` များသည် သင်ကြားရေးအတွင်း စာသားနှင့်ပုံတို့၏ ဆက်နွယ်မှုကို သင်ယူပြီး ေလ့ကျင့်ပြီးနောက် မော်ဒယ်ကို အသုံးပြု၍ အမှားလုံးဖြစ်သည့် အညစ်အကြေးများကို စဉ်ဆက်မပြတ်ပုံတစ်ပုံသို့ ဖော်ဆောင်ပေးသည်။

ပြီးသား နာမည်ကြီး ပုံဖန်တီးမော်ဒယ် မိသားစု ၂ မျိုးမှာ -

- **`gpt-image` (OpenAI)** — ဒီသင်ခန်းစာတွင် အသုံးပြုသော လက်ရှိ မော်ဒယ်မျိုး ဖြစ်သည်။ စာသားမှပုံဖန်တီးခြင်းနှင့် ပုံတည်းဖြတ်ခြင်း (မက့်စ်ဖြင့် အင်ပင်တေးရှင်း) ကို ထောက်ပံ့သည်။
- **Midjourney** — ၎င်း၏ တာဝန်ခံ ဝန်ဆောင်မှုနှင့် Discord အခြေခံ လုပ်ငန်းစဉ်ရှိ ထင်ရှားသော မော်ဒယ်တစ်ခုဖြစ်သည်။

> ဗဟို OpenAI ပုံဖန်တီးမော်ဒယ်များ — **DALL·E 2** နှင့် **DALL·E 3** — သည် အရင်က မော်ဒယ်များ ဖြစ်ကြသည်။ DALL·E 3 သည် အသစ် ထုတ်ဝေမှုအတွက် အသုံးမပြုနိုင်တော့၊ `create_variation` ကဲ့သို့သော လက္ခဏာများသည် DALL·E 2 တွင်ပင်သာ ရှိသည်။ အသစ်အက်ပ်ဖွံ့ဖြိုးရေးအတွက် `gpt-image` မော်ဒယ်များကို အသုံးပြုပါ။

> **အရေးကြီးချက်**: `gpt-image` မော်ဒယ်များသည် ဖန်တီးသော ပုံကို **base64** (`b64_json`) အနေနှင့်သာ ပြန်လည်ပေးပို့ပြီး URL အနေနှင့် မပေးပါ။ သင့်အက်ပ်က base64 စာကြောင်းကို ဒစ်ဂျစ်တယ်ဒေတာအဖြစ် decode လုပ်၍ သိမ်းဆည်းရမည် — ပုံ URL မရှိပါ။


## သင်၏ ပထမဆုံး ပုံဖန်တီးမှု အက်ပ်လီကေးရှင်း တည်ဆောက်ခြင်း

ဒါဆို ပုံဖန်တီးမှု အက်ပ်လီကေးရှင်း တည်ဆောက်ဖို့ ဘာတွေလိုမလဲ? သင်အောက်ပါ ไลဘရဲရီတွေ လိုအပ်မှာဖြစ်ပါတယ် -

- **python-dotenv**, secret များကို ကုဒ်ကနေခွဲထားပြီး *.env* ဖိုင်ထဲမှာ သိမ်းဆည်းဖို့ အကြံပြုအပ်ပါတယ်။
- **openai**, OpenAI API နဲ့ ဆက်သွယ်ဖို့ သုံးမယ့် ไลဘရဲရီဖြစ်ပါတယ်။
- **pillow**, Python မှာ ပုံများနှင့် အလုပ်လုပ်ဖို့။


1. အောက်ပါ အကြောင်းအရာဖြင့် *.env* ဖိုင် တစ်ခု ဖန်တီးပါ -

    ```text
    OPENAI_API_KEY='<add your OpenAI key here>'
    ```



1. အထက်ပါ 라이ဗာရီများကို *requirements.txt* ဟုခေါ်သော ဖိုင်တစ်ခုတွင် အောက်ပါအတိုင်းစုစည်းပါ။

    ```text
    python-dotenv
    openai
    pillow
    ```


1. အောက်တိုင်၊ virtual environment ကိုပြုလုပ်၍ 라이ဗာရီများကို ထည့်သွင်းပါ။


`````python
# create virtual env
! python3 -m venv venv
# activate environment
! source venv/bin/activate
# install libraries
# pip install -r requirements.txt, if using a requirements.txt file 
! pip install python-dotenv openai pillow
`````

> [!NOTE]
> Windows အတွက် သင့်ရဲ့ virtual environment ကို ဖန်တီးပြီး သုံးစွဲဖို့ အောက်ပါ command များကို အသုံးပြုပါ။

    ```bash
    python3 -m venv venv
    venv\Scripts\activate.bat
    ````

1. *app.py* ဟုခေါ်သော ဖိုင်ထဲတွင် အောက်ပါကုဒ်ကို ထည့်ပါ။

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    import openai

    # import dotenv
    dotenv.load_dotenv()

    # Create OpenAI object (reads OPENAI_API_KEY from your .env)
    client = openai.OpenAI()


    try:
        # Create an image by using the image generation API
        generation_response = client.images.generate(
            model="gpt-image-2",
            prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # Enter your prompt text here
            size='1024x1024',
            n=1
        )
        # Set the directory for the stored image
        image_dir = os.path.join(os.curdir, 'images')

        # If the directory doesn't exist, create it
        if not os.path.isdir(image_dir):
            os.mkdir(image_dir)

        # Initialize the image path (note the filetype should be png)
        image_path = os.path.join(image_dir, 'generated-image.png')

        # gpt-image models return the image as base64 (b64_json), not a URL
        image_b64 = generation_response.data[0].b64_json
        generated_image = base64.b64decode(image_b64)
        with open(image_path, "wb") as image_file:
            image_file.write(generated_image)

        # Display the image in the default image viewer
        image = Image.open(image_path)
        image.show()

    # catch exceptions
    except openai.BadRequestError as err:
        print(err)

    ```

ဒီကုဒ်ကို ရှင်းပြကြမယ်။

- ပထမဦးဆုံး ကျွန်တော်တို့လိုအပ်တဲ့ လိုင်ဘရေးရီများကို Import လုပ်ပါတယ်၊ OpenAI 라이ဘရရီ, dotenv 라이ဘရရီ, base64 module နဲ့ Pillow 라이ဘရီအပါအဝင်။

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    import openai
    ```

- ထို့နောက် client ကိုဖန်တီးပါတယ်၊ ဒါက သင့် ``.env`` မှ API key ကိုဖတ်ပါတယ်။

    ```python
    # Create OpenAI object
    client = openai.OpenAI()
    ```

- နောက်တစ်ခုက ပုံကို ထုတ်လုပ်ခြင်း ဖြစ်ပါတယ်။

    ```python
    generation_response = client.images.generate(
        model="gpt-image-2",
        prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # Enter your prompt text here
        size='1024x1024',
        n=1
    )
    ```

    `gpt-image` models တွေအများကြီးပုံကို **base64** string အဖြစ် `data[0].b64_json` မှာ ပြန်တမ်းပေးပါတယ်။ ကျွန်တော်တို့က ဒါကို bytes အဖြစ် decode လုပ်ပြီး ဖိုင်ထဲ သူရိုက်ပါတယ် — ဒါဟာ download လုပ်ဖို့ URL မရှိပါဘူး။

    ```python
    image_b64 = generation_response.data[0].b64_json
    generated_image = base64.b64decode(image_b64)
    with open(image_path, "wb") as image_file:
        image_file.write(generated_image)
    ```

- နောက်ဆုံးမှာ ပုံကိုဖွင့်ပြီး စံပုံမှန် image viewer ကိုသုံးပြီး ပြပါမယ်။

    ```python
    image = Image.open(image_path)
    image.show()
    ```

### ပုံထုတ်လုပ်ခြင်းအကြောင်း အသေးစိတ်

`images.generate` ရဲ့ parameters များကို ကြည့်လိုက်ကြရအောင်။

- **model**, သုံးမယ့် image model ဖြစ်ပါတယ် (ဥပမာ `gpt-image-2`)။
- **prompt**, ပုံထုတ်လုပ်ဖို့ သုံးတဲ့ စာသား prompt ဖြစ်ပါတယ်။ ဒီမှာက "Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils" ဖြစ်ပါတယ်။
- **size**, ထုတ်လုပ်မယ့် ပုံရဲ့ အရွယ်အစားဖြစ်ပါတယ် (ဥပမာ `1024x1024`, `1536x1024`, `1024x1536`, သို့မဟုတ် `"auto"`)။
- **n**, ထုတ်လုပ်မယ့် ပုံအရေအတွက် ဖြစ်ပါတယ်။ ဒီမှာ တစ်ပုံ ထုတ်လုပ်ပါတယ်။

> ပုံထုတ်လုပ်တဲ့ models တွေမှာ `temperature` parameter မရှိပါ — ဒါဟာ စာသား ထုတ်လုပ်ဆဲကို ထိန်းချုပ်ဖို့ ပြုတာဖြစ်ပါတယ်။ အမျိုးမျိုးရဖို့ API ကို ထပ်ခေါ်ပါ။ အမျိုးမျိုးနည်းဖို့ သင့် prompt ကို ပိုသေချာစေပါ။

## ပုံထုတ်လုပ်ခြင်းရဲ့ အပိုဆောင်း အင်္ဂါရပ်များ

Python နှင့် လိုင်းအနည်းငယ်နဲ့ပုံထုတ်လုပ်နိုင်ခြင်းကို ကြည့်ရှုလိုက်ပြီ။ `gpt-image` models တွေကလည်း ရှိပြီးသား ပုံတစ်ပုံကို **တည်းဖြတ်** လုပ်နိုင်ပါတယ်။ ပုံတစ်ပုံ၊ optional **mask** (ပြောင်းလဲသင့်တဲ့ နေရာကို အမှတ်အသားသတ်မှတ်သော) နဲ့ prompt တစ်ခု ပေးပြီး ပုံတစ်ပိုင်းကို ပြောင်းလဲနိုင်ပါတယ် — ဥပမာ ကျွန်တော်တို့ရဲ့ Bunny ပေါ်မှာ hatt တစ်ခု ထည့်နိုင်ပါတယ်။

```python
response = client.images.edit(
  model="gpt-image-2",
  image=open("base_image.png", "rb"),
  mask=open("mask.png", "rb"),
  prompt="An image of a rabbit with a hat on its head.",
  n=1,
  size="1024x1024"
)
# edits are also returned as base64
edited_image = base64.b64decode(response.data[0].b64_json)
```

အခြေခံပုံမှာက ကြွက်ကလေးပဲ ပါတယ်။ နောက်ဆုံးပုံမှာတော့ hatt ပါပြီး ကြွက်ကလေးပေါ်မှာရှိတယ်။

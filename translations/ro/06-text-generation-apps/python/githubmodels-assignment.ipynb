## Ce este o aplicație de generare text?

Normal, când construiești o aplicație, aceasta are un fel de interfață precum următoarea:

- Bazată pe comandă. Aplicațiile console sunt aplicații tipice unde tastezi o comandă și aceasta execută o sarcină. De exemplu, `git` este o aplicație bazată pe comandă.
- Interfață utilizator (UI). Unele aplicații au interfețe grafice (GUI) unde dai clic pe butoane, introduci text, selectezi opțiuni și altele.

### Aplicațiile Console și UI sunt limitate

Compară-l cu o aplicație bazată pe comandă unde tastezi o comandă:

- **Este limitată**. Nu poți introduce orice comandă, doar pe cele pe care aplicația le suportă.
- **Specifică limbajului**. Unele aplicații suportă multe limbi, dar implicit aplicația este construită pentru un anumit limbaj, chiar dacă poți adăuga suport pentru mai multe limbi.

### Beneficiile aplicațiilor de generare text

Atunci cum este diferită o aplicație de generare text?

Într-o aplicație de generare text, ai mai multă flexibilitate, nu ești limitat la un set de comenzi sau un anumit limbaj de intrare. În schimb, poți folosi limbaj natural pentru a interacționa cu aplicația. Un alt beneficiu este că deja interacționezi cu o sursă de date antrenată pe un vast corpus de informații, în timp ce o aplicație tradițională ar putea fi limitată la ceea ce este într-o bază de date.

### Ce pot construi cu o aplicație de generare text?

Sunt multe lucruri pe care le poți construi. De exemplu:

- **Un chatbot**. Un chatbot care răspunde la întrebări despre subiecte, cum ar fi compania ta și produsele acesteia, ar putea fi o alegere bună.
- **Asistent**. Modelele de limbaj mari (LLM) sunt excelente la lucruri precum rezumarea textului, obținerea de perspective din text, generarea de texte cum ar fi CV-uri și altele.
- **Asistent de cod**. În funcție de modelul de limbaj pe care îl folosești, poți construi un asistent de cod care să te ajute să scrii cod. De exemplu, poți folosi produse precum GitHub Copilot, precum și ChatGPT pentru a te ajuta să scrii cod.

## Exercițiu - construiește un generator de rețete

Acum că am parcurs un scenariu, să scriem cod pentru a se potrivi scenariului demonstrat. Pentru aceasta, urmează pașii:

1. Folosește fișierul existent ca punct de plecare
1. Creează o variabilă `prompt` și modifică codul de exemplu așa cum este mai jos:


````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

prompt = "Show me 5 recipes for a dish with the following ingredients: chicken, potatoes, and carrots. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````

Grozav, să vedem cum putem îmbunătăți lucrurile. Pentru a îmbunătăți lucrurile, vrem să ne asigurăm că codul este flexibil, astfel încât ingredientele și numărul de rețete să poată fi îmbunătățite și modificate.

````python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential

# Get these from your Microsoft Foundry project's "Overview" page
token = os.environ["AZURE_INFERENCE_CREDENTIAL"]
endpoint = os.environ["AZURE_INFERENCE_ENDPOINT"]

# temperature/top_p need a non-reasoning model (gpt-5 rejects them), so use a Llama model here.
model_name = os.environ.get("AZURE_INFERENCE_CHAT_MODEL", "Llama-3.3-70B-Instruct")

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)

no_recipes = input("No of recipes (for example, 5): ")

ingredients = input("List of ingredients (for example, chicken, potatoes, and carrots): ")

# interpolate the number of recipes into the prompt an ingredients
prompt = f"Show me {no_recipes} recipes for a dish with the following ingredients: {ingredients}. Per recipe, list all the ingredients used"

response = client.complete(
    messages=[
        {
            "role": "system",
            "content": "You are a helpful assistant.",
        },
        {
            "role": "user",
            "content": prompt,
        },
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.    
)

print(response.choices[0].message.content)
````
